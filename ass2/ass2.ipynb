```python
import pandas as pd

# 1. download and Load the dataset
# load the dataset from train.csv
df = pd.read_csv('train.csv')

# Create a DataFrame and Display the first rows
print("first rows of the dataset")
display(df.head())

# 2. explore the dataset
# determine number of rows and columns
print(f"\ndataset shape (rows, columns): {df.shape}")

# identify data types and presence of missing values 
print("\ndataset info & missing values")
df.info()
print(df.isnull().sum())

# identify numerical and categorical features
# numerical: PassengerId, Survived, Pclass, Age, SibSp, Parch, Fare
# categorical: Sex, Embarked, Name, Ticket, Cabin

# 3. data cleaning (Task B: Median)
# identify missing values before cleaning 
print(f"\nMissing Age values before: {df['Age'].isnull().sum()}")

# remove irrelevant columns: Name, Ticket, Cabin
df.drop(['Name', 'Ticket', 'Cabin'], axis=1, inplace=True)

# Task B: fill missing values using median
df['Age'] = df['Age'].fillna(df['Age'].median())

# remove any remaining rows with missing values (e.g., Embarked)
df.dropna(subset=['Embarked'], inplace=True)

# verify that missing values are handled 
print(f"missing age values after: {df['Age'].isnull().sum()}")
print("\ntotal missing values in final dataset:", df.isnull().sum().sum())

# 4. data transformation
# convert categorical features (Sex, Embarked) into numerical format 
# Map Sex: male to 0, female to 1
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# use One-Hot Encoding for Embarked to ensure only numerical values 
df = pd.get_dummies(df, columns=['Embarked'])

# 5. feature selection
# 5.1. Define the target variable (Survived) 
target = df['Survived']

# 5.2. select at least 4 features 
# chosen features: Pclass, Sex, Age, Fare
features = df[['Pclass', 'Sex', 'Age', 'Fare']]

# 5.3. justification:
# - Sex and Age: 'Women and children first' policy prioritized these groups.
# - Pclass and Fare: Represent socio-economic status and proximity to lifeboats.

# 6. Final Dataset
# ensure no missing values and all features are numerical 
print("\n Final Dataset Types")
print(features.dtypes)
print("\nFinal dataset is ready for model training.")
```

```python
In this assignment I completed the main steps of the machine learning workflow, including data loading, exploration, cleaning, transformation, and feature selection.

The most challenging part was handling missing values and deciding which columns to remove or keep.

I used median imputation for the Age column, removed irrelevant features, and converted categorical variables into numerical format.

The final dataset contains no missing values and only numerical features, so it is ready for machine learning model training.
```
